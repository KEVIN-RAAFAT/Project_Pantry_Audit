```python
def walk(obj, depth=0):
    prefix = "  " * depth
    if isinstance(obj, dict):
        for k, v in obj.items():
            print(f"{prefix}{k}: {type(v).__name__}")
            walk(v, depth + 1)
    elif isinstance(obj, list):
        print(f"{prefix}[list of {len(obj)}]")
        if obj:
            walk(obj[0], depth + 1)

walk(products[0])
```
